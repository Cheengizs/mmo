### Функции для лабы

```python
# определяет гласнная буква чи не
def isVowel(x):
    x = x.lower()
    if (x == 'a' or x == 'e' or
            
        x == 'i' or x == 'o' or x == 'u'):
        return True
    else:
        return False
```

### Задание 2
Посчитать, сколько пар (стоят рядом) верхнего и нижнего регистра
находится в веденном с клавиатуры слове. (Пример HjkLM- 1 пара
нижнего, 1 пара верхнего), а также сколько согласных букв в слове. 

```python
s = input("enter string")
sumUpper = 0
sumLower = 0
consonantAmount = isVowel(s[0]) if 0 else 1
for i in range(1, len(s)):
    if s[i-1].isupper() and s[i].isupper() :
        sumUpper += 1
    elif s[i-1].islower() and s[i].islower():
        sumLower += 1
    if not isVowel(s[i]):
       consonantAmount += 1
print("end upper =", sumUpper)
print("end lower =", sumLower)
print("consonant amount =", consonantAmount)
```

### Задание 5
Реализуйте программу «Кондитерская», которая будет включать в себя шесть пунктов меню. У вас есть словарь, где ключ – название продукции (торт, пирожное, маффин и т.д.). Значение – список, который содержит состав, цену (за 100гр) и кол-во (в граммах).
1. Просмотр описания: название – описание
2. Просмотр цены: название – цена.
3. Просмотр количества: название – количество.
4. Всю информацию.
5. Покупка
В пункте «Покупка» необходимо совершить покупку, с клавиатуры вводите название продукции и его кол-во, n – выход из программы. Посчитать цену выбранных товаров и сколько товаров осталось в изначальном списке
6. До свидания

```python
bakery = {
    "Торт": ["Бисквитные коржи, крем, шоколадная глазурь", 120, 1500],
    "Пирожное": ["Песочное тесто, заварной крем", 80, 800],
    "Маффин": ["Шоколадное тесто, кусочки шоколада", 60, 600],
    "Круассан": ["Слоеное тесто, миндальная начинка", 70, 500]
}

def print_menu():
    print("\n--- Меню «Кондитерская» ---")
    print("1. Просмотр описания")
    print("2. Просмотр цены (за 100г)")
    print("3. Просмотр количества")
    print("4. Вся информация")
    print("5. Покупка")
    print("6. До свидания")

while True:
    print_menu()
    choice = input("Выберите пункт меню (1-6): ").strip()
    
    if choice == "1":
        print("\n--- Описание продукции ---")
        for name, info in bakery.items():
            print(f"{name} – {info[0]}")
            
    elif choice == "2":
        print("\n--- Цены за 100 грамм ---")
        for name, info in bakery.items():
            print(f"{name} – {info[1]} руб.")
            
    elif choice == "3":
        print("\n--- Количество в наличии ---")
        for name, info in bakery.items():
            print(f"{name} – {info[2]} г")
            
    elif choice == "4":
        print("\n--- Полная информация о продукции ---")
        for name, info in bakery.items():
            print(f"{name}: Описание: {info[0]}, Цена за 100г: {info[1]} руб., В наличии: {info[2]} г")
            
    elif choice == "5":
        print("\n--- Режим покупки ---")
        print("(Введите 'n' в качестве названия для завершения покупок)")
        total_bill = 0.0
        
        while True:
            item_name = input("\nВведите название продукции (или 'n' для выхода): ").strip()
            if item_name.lower() == "n":
                break
                
            found_key = None
            for key in bakery:
                if key.lower() == item_name.lower():
                    found_key = key
                    break
                    
            if not found_key:
                print(f"Товар '{item_name}' не найден в кондитерской.")
                continue
                
            qty_str = input(f"Введите количество в граммах для '{found_key}': ").strip()
            if not qty_str.isdigit():
                print("Ошибка: введите целое положительное число грамм.")
                continue
                
            qty = int(qty_str)
            stock = bakery[found_key][2]
            
            if qty <= 0:
                print("Количество должно быть больше 0.")
                continue
            if qty > stock:
                print(f"Недостаточно товара. Доступно только {stock} г.")
                continue
                
            cost = (qty / 100) * bakery[found_key][1]
            total_bill += cost
            bakery[found_key][2] -= qty
            print(f"Добавлено: {found_key} ({qty} г) на сумму {cost:.2f} руб. (Остаток: {bakery[found_key][2]} г)")
            
        print("\n--- Итоги покупки ---")
        print(f"Общая стоимость выбранных товаров: {total_bill:.2f} руб.")
        print("\nОстаток товаров в кондитерской:")
        for name, info in bakery.items():
            print(f"{name}: осталось {info[2]} г")
            
    elif choice == "6":
        print("До свидания! Ждём вас снова в нашей кондитерской.")
        break
    else:
        print("Неверный пункт меню. Пожалуйста, введите число от 1 до 6.")
```
