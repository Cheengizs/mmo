### Задание 1
Создать класс List (список), в котором реализовать методы для работы со
списком (не менее 5). 

```python
class cheengizsList:
    value = None
    nextElem = None

    def __init__(self, *elements):
        self.value = None
        self.nextElem = None
        for el in elements:
            self.append(el)

    def append(self, val):
        if self.value is None and self.nextElem is None:
            self.value = val
            return
        curr = self
        while curr.nextElem is not None:
            curr = curr.nextElem
        new_node = cheengizsList()
        new_node.value = val
        curr.nextElem = new_node

    def prepend(self, val):
        if self.value is None and self.nextElem is None:
            self.value = val
            return
        new_node = cheengizsList()
        new_node.value = self.value
        new_node.nextElem = self.nextElem
        self.value = val
        self.nextElem = new_node

    def remove(self, val):
        if self.value is None and self.nextElem is None:
            raise ValueError(f"{val} не найден в списке")
        if self.value == val:
            if self.nextElem is not None:
                self.value = self.nextElem.value
                self.nextElem = self.nextElem.nextElem
            else:
                self.value = None
            return True
        curr = self
        while curr.nextElem is not None:
            if curr.nextElem.value == val:
                curr.nextElem = curr.nextElem.nextElem
                return True
            curr = curr.nextElem
        raise ValueError(f"{val} не найден в списке")

    def get(self, index):
        if index < 0:
            raise IndexError("Индекс не может быть отрицательным")
        curr = self
        idx = 0
        while curr is not None and (curr.value is not None or curr.nextElem is not None):
            if idx == index:
                return curr.value
            curr = curr.nextElem
            idx += 1
        raise IndexError("Индекс выходит за границы списка")

    def find(self, val):
        curr = self
        idx = 0
        while curr is not None and (curr.value is not None or curr.nextElem is not None):
            if curr.value == val:
                return idx
            curr = curr.nextElem
            idx += 1
        return -1

    def size(self):
        if self.value is None and self.nextElem is None:
            return 0
        count = 0
        curr = self
        while curr is not None and (curr.value is not None or curr.nextElem is not None):
            count += 1
            curr = curr.nextElem
        return count

    def clear(self):
        self.value = None
        self.nextElem = None

    def to_list(self):
        res = []
        if self.value is None and self.nextElem is None:
            return res
        curr = self
        while curr is not None and (curr.value is not None or curr.nextElem is not None):
            res.append(curr.value)
            curr = curr.nextElem
        return res

    def __len__(self):
        return self.size()

    def __str__(self):
        lst = self.to_list()
        return " -> ".join(map(str, lst)) if lst else "Пустой список"

my_list = cheengizsList(10, 20, 30)
print("1. Создан список:", my_list)

my_list.append(40)
print("2. Добавили 40 в конец:", my_list)

my_list.prepend(5)
print("3. Добавили 5 в начало:", my_list)

print("5. Элемент по индексу 3:", my_list.get(3))
print("6. Индекс элемента 30:", my_list.find(30))

my_list.remove(20)
print("7. Удалили элемент 20:", my_list)

print("9. Размер списка:", len(my_list))
print("10. Обычный список в Python:", my_list.to_list())
```

### Задание 2
Создать классы «Транспортное средство», «Самолет», «Поезд»,
«Автомобиль». Определить время и стоимость перевозки для указанных
городов и расстояний. Вывести данные о наиболее быстрой и экономичной
поездке. Предусмотреть метод записи информации в файл

```python

class Vehicle:
    def __init__(self, name, speed, cost_per_km):
        self.name = name                      
        self.speed = speed                    
        self.cost_per_km = cost_per_km        

    def calculate_time(self, distance):
        """Время поездки в часах: t = S / v"""
        return distance / self.speed

    def calculate_cost(self, distance):
        """Стоимость поездки: C = S * cost_per_km"""
        return distance * self.cost_per_km

    def get_trip_info(self, from_city, to_city, distance):
        time_hours = self.calculate_time(distance)
        cost = self.calculate_cost(distance)
        return {
            "vehicle": self.name,
            "route": f"{from_city} -> {to_city}",
            "distance_km": distance,
            "time_hours": round(time_hours, 2),
            "cost_rub": round(cost, 2)
        }

    def save_info_to_file(self, filename, from_city, to_city, distance):
        """Метод записи информации о поездке конкретного транспорта в файл"""
        info = self.get_trip_info(from_city, to_city, distance)
        with open(filename, 'a', encoding='utf-8') as f:
            f.write(f"{info['vehicle']} | Маршрут: {info['route']} ({info['distance_km']} км) | "
                    f"Время: {info['time_hours']} ч. | Стоимость: {info['cost_rub']} руб.\n")


class Airplane(Vehicle):
    def __init__(self, name="Самолет", speed=800, cost_per_km=12.0):
        super().__init__(name, speed, cost_per_km)

class Train(Vehicle):
    def __init__(self, name="Поезд", speed=100, cost_per_km=3.5):
        super().__init__(name, speed, cost_per_km)

class Car(Vehicle):
    def __init__(self, name="Автомобиль", speed=90, cost_per_km=5.5):
        super().__init__(name, speed, cost_per_km)


ТранспортноеСредство = Vehicle
Самолет = Airplane
Поезд = Train
Автомобиль = Car

def analyze_trips(vehicles, from_city, to_city, distance, output_file="trip_report.txt"):
    print(f"\n=== Маршрут: {from_city} — {to_city} ({distance} км) ===")
    trips = [v.get_trip_info(from_city, to_city, distance) for v in vehicles]
    
    for t in trips:
        print(f"{t['vehicle']}: Время = {t['time_hours']} ч, Стоимость = {t['cost_rub']} руб.")
    
    fastest = min(trips, key=lambda x: x["time_hours"])
    cheapest = min(trips, key=lambda x: x["cost_rub"])
    
    print("\n>> Итоги анализа:")
    print(f"Самая быстрая поездка: {fastest['vehicle']} ({fastest['time_hours']} ч)")
    print(f"Самая экономичная поездка: {cheapest['vehicle']} ({cheapest['cost_rub']} руб.)")
    
    with open(output_file, "w", encoding="utf-8") as f:
        f.write(f"Отчет о поездках по маршруту {from_city} — {to_city} ({distance} км)\n")
        f.write("=" * 60 + "\n")
        for t in trips:
            f.write(f"{t['vehicle']}: время = {t['time_hours']} ч, стоимость = {t['cost_rub']} руб.\n")
        f.write("=" * 60 + "\n")
        f.write(f"Наиболее быстрая поездка: {fastest['vehicle']} ({fastest['time_hours']} ч)\n")
        f.write(f"Наиболее экономичная поездка: {cheapest['vehicle']} ({cheapest['cost_rub']} руб.)\n")
    
    print(f"\nРезультаты успешно сохранены в файл: {output_file}")

vehicles_list = [Airplane(), Train(), Car()]

analyze_trips(vehicles_list, "Минск", "Москва", 720)
```

### Задание 3
Создать классы: «Книга», «Отдел», «Библиотека». В классах реализовать
следующие методы: добавление, удаление, изменение названия книг из
отделов. Классы должны содержать методы доступа и изменения всех поле.
Предусмотреть метод записи информации в файл.

```python
class Book:
    def __init__(self, title, author, year):
        self._title = title
        self._author = author
        self._year = year

    @property
    def title(self):
        return self._title

    @title.setter
    def title(self, new_title):
        if not new_title:
            raise ValueError("Название книги не может быть пустым.")
        self._title = new_title

    @property
    def author(self):
        return self._author

    @author.setter
    def author(self, new_author):
        self._author = new_author

    @property
    def year(self):
        return self._year

    @year.setter
    def year(self, new_year):
        self._year = new_year

    def get_title(self):
        return self._title

    def set_title(self, new_title):
        self.title = new_title

    def get_author(self):
        return self._author

    def set_author(self, new_author):
        self._author = new_author

    def get_year(self):
        return self._year

    def set_year(self, new_year):
        self._year = new_year

    def __str__(self):
        return f"«{self._title}» (Автор: {self._author}, Год: {self._year})"


class Department:
    def __init__(self, name):
        self._name = name
        self._books = []

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        self._name = new_name

    @property
    def books(self):
        return self._books

    def get_name(self):
        return self._name

    def set_name(self, new_name):
        self._name = new_name

    def add_book(self, book):
        if not isinstance(book, Book):
            raise TypeError("Ожидается объект класса Book")
        self._books.append(book)
        print(f"Книга {book} добавлена в отдел '{self._name}'")

    def remove_book(self, title):
        for i, book in enumerate(self._books):
            if book.title.lower() == title.lower():
                removed = self._books.pop(i)
                print(f"Книга {removed} удалена из отдела '{self._name}'")
                return True
        print(f"Книга с названием '{title}' не найдена в отделе '{self._name}'")
        return False

    def change_book_title(self, old_title, new_title):
        for book in self._books:
            if book.title.lower() == old_title.lower():
                book.title = new_title
                print(f"В отделе '{self._name}' название книги изменено с '{old_title}' на '{new_title}'")
                return True
        print(f"Книга с названием '{old_title}' не найдена для изменения")
        return False

    def find_book(self, title):
        for book in self._books:
            if book.title.lower() == title.lower():
                return book
        return None

    def __str__(self):
        result = f"Отдел: {self._name} (Книг: {len(self._books)})\n"
        for i, b in enumerate(self._books, 1):
            result += f"   {i}. {b}\n"
        return result

class Library:
    def __init__(self, name):
        self._name = name
        self._departments = {}

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        self._name = new_name

    def get_name(self):
        return self._name

    def set_name(self, new_name):
        self._name = new_name

    def add_department(self, dept):
        if not isinstance(dept, Department):
            raise TypeError("Ожидается объект класса Department")
        self._departments[dept.name] = dept
        print(f"Отдел '{dept.name}' добавлен в библиотеку '{self._name}'")

    def get_department(self, dept_name):
        return self._departments.get(dept_name)

    def add_book_to_department(self, dept_name, book):
        dept = self.get_department(dept_name)
        if dept:
            dept.add_book(book)
        else:
            print(f"Отдел '{dept_name}' не найден в библиотеке.")

    def remove_book_from_department(self, dept_name, book_title):
        dept = self.get_department(dept_name)
        if dept:
            dept.remove_book(book_title)
        else:
            print(f"Отдел '{dept_name}' не найден в библиотеке.")

    def change_book_title_in_department(self, dept_name, old_title, new_title):
        dept = self.get_department(dept_name)
        if dept:
            dept.change_book_title(old_title, new_title)
        else:
            print(f"Отдел '{dept_name}' не найден в библиотеке.")

    def save_to_file(self, filename="library_data.txt"):
        with open(filename, "w", encoding="utf-8") as f:
            f.write(f"=== Библиотека: {self._name} ===\n\n")
            for dept_name, dept in self._departments.items():
                f.write(str(dept) + "\n")
        print(f"Данные библиотеки успешно записаны в файл: {filename}")

    def display_catalog(self):
        print(f"\n{'=' * 40}")
        print(f"Каталог библиотеки «{self._name}»")
        print(f"{'=' * 40}")
        if not self._departments:
            print("В библиотеке пока нет отделов.")
        for dept in self._departments.values():
            print(dept)

Книга = Book
Отдел = Department
Библиотека = Library

lib = Library("Национальная библиотека")

dept_fiction = Department("Художественная литература")
dept_science = Department("Научная литература")
lib.add_department(dept_fiction)
lib.add_department(dept_science)

lib.add_book_to_department("Художественная литература", Book("Война и мир", "Л. Толстой", 1869))
lib.add_book_to_department("Художественная литература", Book("1984", "Дж. Оруэлл", 1949))
lib.add_book_to_department("Научная литература", Book("Краткая история времени", "С. Хокинг", 1988))

lib.display_catalog()

lib.change_book_title_in_department("Художественная литература", "1984", "1984 (Специальное издание)")

lib.remove_book_from_department("Художественная литература", "Война и мир")

lib.display_catalog()

lib.save_to_file("library_catalog.txt")
```
