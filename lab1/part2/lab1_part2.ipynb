### Задание 1
Напишите функцию is_password_good(password), которая принимает
в качестве аргумента строковое значение пароля password и возвращает
значение True если пароль является надежным и False в противном случае.
Пароль является надежным если:
– его длина не менее 10 символов;
– он содержит как минимум одну заглавную букву (верхний регистр);
– он содержит как минимум одну строчную букву (нижний регистр).

```python
def is_password_good(password):
    try:
        if len(password) < 10:
            return False
        if not any(symb.isupper() for symb in password):
            return False
        if not any(symb.islower() for symb in password):
            return False

        return True
    except TypeError as e:
        print("Ошибка: передан неверный тип данных:", e)
        return False
    except Exception as e:
        print("Произошла непредвиденная ошибка:", e)
        return False
    finally:
        print("Проверка пароля завершена.")

try:
    print(is_password_good('PIUBIUBAFIUeB'))
except Exception as e:
    print("Ошибка вызова:", e)
finally:
    print("Завершение выполнения Задания 1")
```

### Задание 2

функцию передаётся кортеж, то посчитать длину всех его слов.
Если список, то сумму после первого отрицательного элемента.
Удалить все повторяющиеся элементы.
Число – кол-во четных цифр.
Строка – найти сумму всех чисел.
Сделать проверку со всеми этими случаями


```python
def diffTypeFunction(a):
    try:
        if type(a) == tuple:
            print("Tuple")
            for i in a:
                print("len of ", i, "is", len(i))
        if type(a) == list:
            print("list")
            isNegativeFound = False
            sumAfterNegatvie = 0
            for i in range(0, len(a)):
                if isNegativeFound:
                    sumAfterNegatvie += a[i]
                elif a[i] < 0:
                    isNegativeFound = True
            print("sum after negative:", sumAfterNegatvie)
        if type(a) == int:
            print("number")
            numStr = str(a)
            print("amount of even digits:", sum(symb in "02468" for symb in numStr))
        if type(a) == str:
            print("stroka")
            wordNumArr = a.split()
            numberSum = 0
            for word in wordNumArr:
                if(word.isnumeric()):
                    numberSum += int(word)
            print("sum of numbers in string:", numberSum)
    except TypeError as e:
        print("Ошибка типа данных при обработке элемента:", e)
    except Exception as e:
        print("Произошла непредвиденная ошибка:", e)
    finally:
        print("Обработка аргумента завершена.")

try:
    diffTypeFunction(("apple", "banana", "kiwi"))
    diffTypeFunction([1, 2, -3, 4, 5])
    diffTypeFunction(24681)
    diffTypeFunction("some 10 20 words")
except Exception as e:
    print("Ошибка при тестировании функции:", e)
finally:
    print("Завершение выполнения Задания 2")
```

### Задание 3
Дан двумерный массив и два числа: i и j. Поменяйте в массиве
столбцы с номерами i и j и выведите результат

```python
def changeIandJcolumnsInMatrix(matrix, i, j):
    try:
        for row in matrix:
            (row[i - 1], row[j - 1]) = (row[j - 1], row[i - 1])
    except IndexError as e:
        print("Ошибка: индекс столбца выходит за границы матрицы:", e)
    except Exception as e:
        print("Ошибка при перестановке столбцов:", e)
    finally:
        print("Операция перестановки столбцов завершена.")

try:
    s = input("input n and m")
    n, m = map(int, s.split())

    matrix = []

    for i in range(0, n):
        row = list(map(int, input().split()))
        matrix.append(row)

    for row in matrix:
        print(row)

    changeIandJcolumnsInMatrix(matrix, 1, 3)

    for row in matrix:
        print(row)
except ValueError as e:
    print("Ошибка ввода данных (ValueError):", e)
except IndexError as e:
    print("Ошибка индекса матрицы (IndexError):", e)
except Exception as e:
    print("Произошла ошибка:", e)
finally:
    print("Завершение выполнения Задания 3")
```

### Задание 4
Напишите программу, демонстрирующую работу try\except\finally

#### Уже сделано в прошлых тасках

```python
# Задание 4: Демонстрация работы try\except\else\finally
def process_data(value):
    try:
        print(f"\nОбработка значения: {repr(value)}")
        num = float(value)
        result = 100 / num
    except ValueError as e:
        print(f"  [except ValueError]: Ошибка преобразования ({e})")
    except ZeroDivisionError as e:
        print(f"  [except ZeroDivisionError]: Деление на ноль ({e})")
    else:
        print(f"  [else]: Успешный расчет, результат = {result:.2f}")
    finally:
        print("  [finally]: Блок гарантированного завершения выполнен.")

for val in ["25", "0", "abc", 10]:
    process_data(val)
```
